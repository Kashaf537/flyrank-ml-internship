# My Lane: Content Refresh Prioritization

## ML Task Type: Ranking / Scoring

My goal is to prioritize content pages that are most likely to benefit from a refresh. Rather than simply predicting whether a page is declining, I want to produce a ranked list of pages so that a content team can focus its limited time and resources on the highest-priority pages first.

The output would be a priority score for each content page, allowing pages to be ranked from highest to lowest refresh priority.

## 2. Target or proxy

*What would you predict? Where does that label come from — observed outcome or a defined rule?*

My eventual goal is to create a refresh priority score for each content page. The score would represent how strongly the available evidence suggests that a page deserves attention from the content team.

Potential signals include impressions_90d, ctr, engagement_rate, content_age_days, and avg_position.

For the current starter dataset, I will use the observed declining-content label as an exploratory proxy for identifying pages that may require attention. However, I will treat this carefully because the label is derived from trend_direction, and therefore it should not be used as a model feature. A stronger future setup would define the target using a future time window so that the model predicts an outcome that occurs after the features are observed.

## 3. Success metric

*One metric you can defend. What number means 'good'?*

# Success Metric: Precision@K

I will evaluate the ranking using Precision@K, particularly Precision@20 and Precision@50. This measures how many of the top K pages recommended for refresh are actually relevant according to the evaluation target.

This metric matches the business decision because content teams have limited time and cannot refresh every page. The goal is to make the top recommendations as useful as possible.

For example, Precision@20 answers: "Of the 20 pages prioritized for refresh, how many were actually relevant?" A higher Precision@20 means the content team spends less time investigating low-value recommendations.

## 5. Why ML beats a fixed rule here

*What makes the pattern too messy for an if-statement?*

A fixed rule could identify pages using a single threshold, such as low CTR or low impressions. However, content performance is influenced by multiple factors, including impressions, CTR, engagement, content age, and search position. A page may have low CTR but still receive substantial traffic, while an older page may perform well and not need immediate attention.

ML could be useful if these signals interact in complex ways that are difficult to capture with a small set of manually defined rules. A model could learn combinations of signals and produce a more useful ranking of pages for the content team.

However, ML is not automatically better than a rule. I would first compare the ML approach against a simple, transparent baseline and only use ML if it provides a meaningful improvement in ranking quality.
