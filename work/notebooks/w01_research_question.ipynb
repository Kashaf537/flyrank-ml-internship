## My Lane: Content Refresh Prioritization

I selected the **Content Refresh Prioritization** lane because websites often contain hundreds or thousands of pages, making it difficult for content teams to decide which pages should be updated first. My goal is to use the available page performance metrics to identify pages that may require refreshing so editors can focus on the highest-priority content. This problem is suitable for machine learning because multiple factors such as impressions, CTR, rankings, and content age may jointly influence page performance.

## 2. The question: decision, action, cost of a wrong call

*What decision does your work improve? Who acts on it? What does a wrong recommendation cost?*

## Research Question

**Decision**

Which content pages should be refreshed first to maintain or improve search performance?

**Who acts on the output?**

SEO specialists or content editors will use the recommendations.

**Action**

The recommended pages will be reviewed, updated, or optimized before their performance declines further.

**Cost of a wrong recommendation**

If a page is incorrectly identified as needing a refresh, editors may spend time updating content that does not require changes. If a declining page is missed, the website may continue losing traffic and ranking opportunities.

**Why can machine learning help?**

Page performance depends on several metrics rather than a single factor. Machine learning can combine multiple signals to prioritize pages more effectively than a simple manual rule.

```python
import pandas as pd

df = pd.read_csv(
    "flyrank-ml-internship/data/raw/content_refresh_anonymized.csv"
)
print("Total Pages:", len(df))
print("Average CTR:", round(df["ctr"].mean(),3))
print("Average Content Age (days):", round(df["content_age_days"].mean(),1))

print("\nTrend Direction Counts")
print(df["trend_direction"].value_counts())
```

## Observations

-The dataset contains 30,000 content pages, providing a large sample for analysis.

-The average CTR is 0.511%, and the average content age is approximately 256 days, indicating that the dataset contains pages with different levels of search performance and age.

-The trend_direction distribution shows that 16,262 pages are declining, which is the largest category.

-This suggests that prioritizing declining pages could be a useful decision-support problem for content teams.
